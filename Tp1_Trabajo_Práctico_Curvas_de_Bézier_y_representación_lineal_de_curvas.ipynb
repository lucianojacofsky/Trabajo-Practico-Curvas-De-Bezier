 $$x_1(t) = (1-t)(2) + t(4) = 2 - 2t + 4t = 2 + 2t$$

# Ejercicio 3: Construcción de una curva de Bézier cúbica y efecto del orden de los puntos de control

A partir de los puntos anteriores $P_0=(1,1)$, $P_1=(2,4)$, $P_2=(4,0)$ y agregando $P_3=(4,-1)$.



---



### a) Curva de Bézier cúbica (Orden original)
La curva de Bézier cúbica usando los puntos de control en el orden $(P_0, P_1, P_2, P_3)$:

<br>

$$B(t) = (1-t)^3 P_0 + 3(1-t)^2 t P_1 + 3(1-t) t^2 P_2 + t^3 P_3$$



<br>
<br>



### c) Nueva curva de Bézier con nuevo orden: $P_0, P_2, P_1, P_3$
Al cambiar el orden de los puntos internos, la función cambia a:

<br>

$$B^´(t) = (1-t)^3 P_0 + 3(1-t)^2 t P_2 + 3(1-t) t^2 P_1 + t^3 P_3$$



---



```python
import numpy as np
import matplotlib.pyplot as plt

# Puntos originales
P0, P1, P2, P3 = np.array([1,1]), np.array([2,4]), np.array([4,0]), np.array([4,-1])

def bezier_cubica(t, pts):
    p0, p1, p2, p3 = pts
    return (1-t)**3 * p0 + 3*(1-t)**2 * t * p1 + 3*(1-t) * t**2 * p2 + t**3 * p3

t_vals = np.linspace(0, 1, 100)

# Caso 1: Orden Original
curva1 = np.array([bezier_cubica(t, [P0, P1, P2, P3]) for t in t_vals])

# Caso 2: Orden Cambiado (P0, P2, P1, P3)
curva2 = np.array([bezier_cubica(t, [P0, P2, P1, P3]) for t in t_vals])

# Graficar
fig, ax = plt.subplots(1, 2, figsize=(15, 5))

# Gráfico 1
ax[0].plot(curva1[:,0], curva1[:,1], 'b-', label='Bézier Original')
ax[0].plot([P0[0], P1[0], P2[0], P3[0]], [P0[1], P1[1], P2[1], P3[1]], 'ro--')
ax[0].set_title('Orden: P0, P1, P2, P3')

# Gráfico 2
ax[1].plot(curva2[:,0], curva2[:,1], 'g-', label='Bézier Cambiada')
ax[1].plot([P0[0], P2[0], P1[0], P3[0]], [P0[1], P2[1], P1[1], P3[1]], 'ro--')
ax[1].set_title('Orden: P0, P2, P1, P3')

for a in ax:
    a.grid(True)
    a.legend()

plt.show()
```

### e) Justificación
La curva no es la misma. Aunque los puntos en el plano sean los mismos, la curva de Bézier depende de la secuencia. La curva siempre intenta "seguir" esa secuencia de puntos ordenados, en el tiempo $t$.

# Ejercicio 6: Forma matricial de una curva de Bézier cúbica

### a) Escribir $B(t)$ en la forma

$$B(t) = \mathbf{P} \cdot M \cdot \mathbf{T}$$

2. $3(1-t)^2t = 3(1 - 2t + t^2)t = 3t - 6t^2 + 3t^3$
3. $3(1-t)t^2 = 3(t^2 - t^3) = 3t^2 - 3t^3$
4. $t^3 = 0 + 0t + 0t^2 + 1t^3$

Para que el producto $\mathbf{P} \cdot M \cdot \mathbf{T}$ sea equivalente a $B(t) = (1-t)^3 P_0 + 3(1-t)^2 t P_1 + 3(1-t) t^2 P_2 + t^3 P_3$, la matriz $M$ debe contener los coeficientes de cada polinomio en sus **filas** correspondientes(Ej: Fila 1, escalar $P_0$):

$$M = \begin{bmatrix}

0 & 3 & -6 & 3 \\
0 & 0 & 3 & -3 \\
0 & 0 & 0 & 1
\end{bmatrix}$$

<br>
<br>

### c) Verificar el resultado expandiendo el producto matricial
Al multiplicar $M \cdot \mathbf{T}$ =
$\begin{bmatrix}

3t - 6t^2 + 3t^3 \\
3t^2 - 3t^3 \\
t^3
\end{bmatrix}$.

Luego, $\mathbf{P} \cdot M \cdot \mathbf{T}$ = $\begin{bmatrix}

**Verificado:** $B(t) = \begin{bmatrix} P_0 & P_1 & P_2 & P_3 \end{bmatrix} \cdot M \cdot  \begin{bmatrix}

t \\
t^2 \\
t^3
\end{bmatrix}$

<br>

### b) Verificación
Una propiedad clave es que la suma de las funciones base siempre es igual a 1 para cualquier $t$. Lo verificamos expandiendo los términos:
1. $(1-t)^3 = 1 - 3t + 3t^2 - t^3$
2. $3(1-t)^2t = 3t - 6t^2 + 3t^3$
3. $3(1-t)t^2 = 3t^2 - 3t^3$
4. $t^3 = t^3$

Al sumar todo:
$$(1 - 3t + 3t^2 - t^3) + (3t - 6t^2 + 3t^3) + (3t^2 - 3t^3) + (t^3) = 1$$

<br>

Se anulan todos entre ellos excepto el 1 del principio, quedando como resultado.

### c) Calcular
Calculamos los valores de cada función en los bordes del intervalo:

* **En $t = 0$:**
  * $b_{0}(0) = (1-0)^3 = 1$

  * $b_{3}(0) = 0^{3}=0$

* **En $t = 1$:**
  * $b_{0}(1) = (1-1)^3 = 0^{3}=0$
  * $b_{1}(1) = 3(1-1)^{2}1= (0^2)1=0$

  * $b_{3}(1) = 1^{3}=1$

* Como en $t=0$ la única función que no se anula es $b_0(t)$ (y vale 1), la fórmula de la curva se reduce a $B(0) = 1 \cdot P_0$, por lo que la curva **comienza** en $P_0$.
* Como en $t=1$ la única función que no se anula es $b_3(t)$, la fórmula se reduce a $B(1) = 1 \cdot P_3$, por lo que la curva **termina** en $P_3$.

### a) Plantear la ecuación
Para determinar si son linealmente independientes, planteamos una combinación lineal igualada a cero:
$$\alpha(1-t)^3 + \beta \cdot 3(1-t)^2t + \gamma \cdot 3(1-t)t^2 + \delta t^3 = 0$$

<br>

### b) Expandir y reagrupar en la base monomial
Expandimos los términos y agrupamos según las potencias de $t$:
1. $\alpha(1 - 3t + 3t^2 - t^3)=0$
2. $\beta(3t - 6t^2 + 3t^3)=0$
3. $\gamma(3t^2 - 3t^3)=0$
4. $\delta(t^3)=0$


Esto equivale a formar una matriz cuyas columnas sean los resultados anteriormente calculados y resolver el sistema:

$$\begin{bmatrix} 1 & 0 & 0 & 0 \\ -3 & 3 & 0 & 0 \\ 3 & -6 & 3 & 0 \\ -1 & 3 & -3 & 1 \end{bmatrix} \begin{bmatrix} \alpha \\ \beta \\ \gamma \\ \delta \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \\ 0 \\ 0 \end{bmatrix}$$

<br>

### c) Determinar $α, β, γ, δ$.

2. De la segunda fila: $-3\alpha + 3\beta = 0 \implies 3\beta = 0 \implies \beta = 0$
3. De la tercera fila: $3\alpha - 6\beta + 3\gamma = 0 \implies 3\gamma = 0 \implies \gamma = 0$
4. De la cuarta fila: $-\alpha + 3\beta - 3\gamma + \delta = 0 \implies \delta = 0$

<br>

### d) Concluir si $\mathcal{B}$ es linealmente independiente.
Dado que la única solución es la trivial ($\alpha = \beta = \gamma = \delta = 0$), los vectores de coordenadas son linealmente independientes, formando así una base.

# Ejercicio 10: Cambio entre la base monomial y la base de Bernstein

Base monomial $M = \{1, t, t^2, t^3\}$  
Base de Bernstein $B = \{b_{0,3}(t), b_{1,3}(t), b_{2,3}(t), b_{3,3}(t)\}$.

---

### a) Matriz de cambio de base
Para encontrar la matriz que lleva coordenadas de la base de Bernstein a la monomial, expandimos los polinomios de grado 3:
1. $b_{0,3}(t) = (1-t)^3 = 1 - 3t + 3t^2 - t^3$
2. $b_{1,3}(t) = 3t(1-t)^2 = 3t(1 - 2t + t^2) = 3t - 6t^2 + 3t^3$
3. $b_{2,3}(t) = 3t^2(1-t) = 3t^2 - 3t^3$
4. $b_{3,3}(t) = t^3$

Colocando los coeficientes de cada polinomio como columnas, obtenemos la matriz $A$:

$$A = \begin{bmatrix} 1 & 0 & 0 & 0 \\ -3 & 3 & 0 & 0 \\ 3 & -6 & 3 & 0 \\ -1 & 3 & -3 & 1 \end{bmatrix}$$

### b) Cálculo del determinante
La matriz $A$ es una **matriz triangular inferior**. El determinante de una matriz triangular es el producto de los elementos de su diagonal principal:

$$\det(A) = 1 \cdot 3 \cdot 3 \cdot 1 = 9$$

### c) Invertibilidad
Dado que el determinante es $\det(A) = 9$ y, por lo tanto, es **distinto de cero** ($\neq 0$), la matriz es **invertible** (no singular).

### d) Implicación sobre la relación entre las bases
La invertibilidad de esta matriz implica que:
1. **Bases equivalentes:** Tanto la base monomial como la de Bernstein son bases válidas para el mismo espacio vectorial de polinomios de grado menor o igual a 3. Cualquier curva que se pueda expresar en una, se puede expresar en la otra sin pérdida de información.
2. **Transformación Biyectiva:** Existe una correspondencia única entre los puntos de control de una curva de Bézier y los coeficientes de un polinomio tradicional.
3. **Estabilidad:** En términos de computación, esto justifica por qué podemos usar la matriz para cambiar de representación según nos convenga (por ejemplo, la base de Bernstein es mucho más estable numéricamente para diseño gráfico que la base monomial).

# Ejercicio 11: Implementación computacional

A partir de  los puntos: $P_0 = (0,0)$, $P_1 = (2,4)$ y $P_2 = (4,0)$.

---


```python
import numpy as np
import matplotlib.pyplot as plt

# Puntos de control definidos
P0 = np.array([0, 0])
P1 = np.array([2, 4])
P2 = np.array([4, 0])
puntos_control = np.array([P0, P1, P2])

# a) Implementación de la fórmula explícita
def bezier_explicita(t, p0, p1, p2):
    return (1-t)**2 * p0 + 2*(1-t)*t * p1 + t**2 * p2

# d) Implementación de la construcción iterada (Ejercicio 2)
def bezier_iterada(t, p0, p1, p2):
    q0 = (1-t)*p0 + t*p1
    q1 = (1-t)*p1 + t*p2
    return (1-t)*q0 + t*q1

# b) Evaluación para 100 valores de t
t_100 = np.linspace(0, 1, 100)
curva_puntos = np.array([bezier_explicita(t, P0, P1, P2) for t in t_100])

# c) Graficar puntos de control, poligonal y curva
plt.figure(figsize=(10, 6))
plt.plot(curva_puntos[:, 0], curva_puntos[:, 1], 'b-', label='Curva de Bézier')
plt.plot(puntos_control[:, 0], puntos_control[:, 1], 'ro--', alpha=0.5, label='Poligonal de control')
plt.scatter(puntos_control[:, 0], puntos_control[:, 1], color='red')

# Etiquetas para los puntos
for i, p in enumerate(puntos_control):
    plt.text(p[0], p[1], f' P{i}', verticalalignment='bottom')

plt.title('Ejercicio 11: Curva Cuadrática y Poligonal de Control')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.show()

# Comparación de métodos para 10 valores
t_10 = np.linspace(0, 1, 10)
print(f"{'t':<10} | {'Explicita':<20} | {'Iterada':<20} | {'Diferencia':<15}")
print("-" * 70)

for t in t_10:
    exp = bezier_explicita(t, P0, P1, P2)
    ite = bezier_iterada(t, P0, P1, P2)
    diff = np.linalg.norm(exp - ite)
    print(f"{t:<10.2f} | {str(exp):<20} | {str(ite):<20} | {diff:<15.2e}")
```

### a) Fórmula explícita de $h(t)$
Definimos $g_1(t)$ y $g_2(t)$ como:  
$g_1(t) = (1-t)^2 P_0 + 2(1-t)t P_1 + t^2 P_2$  
$g_2(t) = (1-t)^2 P_1 + 2(1-t)t P_2 + t^2 P_3$

La curva combinada es $h(t) = (1-t)g_1(t) + t g_2(t)$. Sustituyendo:

<br>

$h(t) = (1-t)[(1-t)^2 P_0 + 2(1-t)t P_1 + t^2 P_2] + t[(1-t)^2 P_1 + 2(1-t)t P_2 + t^2 P_3]$



$h(t) = (1-t)^3 P_0 + 2(1-t)^2t P_1 + (1-t)t^2 P_2 + t(1-t)^2 P_1 + 2t^2(1-t) P_2 + t^3 P_3$

<br>

$$h(t) = (1-t)^3 P_0 + 3(1-t)^2t P_1 + 3(1-t)t^2 P_2 + t^3 P_3$$

Esta expresión corresponde exactamente a la definición de una **curva de Bézier cúbica**, donde los coeficientes son los polinomios de Bernstein de grado 3.

```python
import numpy as np
import matplotlib.pyplot as plt

# b) Función h(t)
def h_bezier(t, puntos):
    P0, P1, P2, P3 = puntos
    # Polinomios de Bernstein de grado 3
    b0 = (1-t)**3
    b1 = 3 * (1-t)**2 * t
    b2 = 3 * (1-t) * t**2
    b3 = t**3
    return b0 * P0 + b1 * P1 + b2 * P2 + b3 * P3

# c) Representación de los coeficientes (Polinomios de Bernstein)
t_vals = np.linspace(0, 1, 100)
b0_vals = (1-t_vals)**3
b1_vals = 3 * (1-t_vals)**2 * t_vals
b2_vals = 3 * (1-t_vals) * t_vals**2
b3_vals = t_vals**3

plt.figure(figsize=(10, 5))
plt.plot(t_vals, b0_vals, label='$b_0(t) = (1-t)^3$')
plt.plot(t_vals, b1_vals, label='$b_1(t) = 3t(1-t)^2$')
plt.plot(t_vals, b2_vals, label='$b_2(t) = 3t^2(1-t)$')
plt.plot(t_vals, b3_vals, label='$b_3(t) = t^3$')
plt.title('Ejercicio 12c: Coeficientes de los Puntos de Control (Polinomios de Bernstein)')
plt.xlabel('t')
plt.ylabel('Valor del coeficiente')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### d) Derivadas y Forma Matricial

La forma matricial de la curva es $h(t) = \mathbf{T} \cdot \mathbf{M} \cdot \mathbf{P}$, donde:
$\mathbf{T} = \begin{bmatrix} 1 & t & t^2 & t^3 \end{bmatrix}$, $\mathbf{P} = \begin{bmatrix} P_0 \\ P_1 \\ P_2 \\ P_3 \end{bmatrix}$ y $\mathbf{M} = \begin{bmatrix} 1 & 0 & 0 & 0 \\ -3 & 3 & 0 & 0 \\ 3 & -6 & 3 & 0 \\ -1 & 3 & -3 & 1 \end{bmatrix}$

```python
import numpy as np
import matplotlib.pyplot as plt

# a) Generar puntos al azar y función Bézier
def generar_puntos():
    return np.random.uniform(-10, 10, (4, 2))

def bezier_cubica(t, P):
    return (1-t)**3 * P[0] + 3*(1-t)**2*t * P[1] + 3*(1-t)*t**2 * P[2] + t**3 * P[3]

P = generar_puntos() # Puntos para B1
Q = generar_puntos() # Puntos para B2
```

### b) Demostración analítica
Sea $T(\mathbf{x}) = A\mathbf{x}$ una transformación lineal. Aplicándola a la curva:
$$T(B(t)) = T\left( \sum_{i=0}^3 b_i(t) P_i \right)$$
Por la propiedad de linealidad ($T(u+v) = T(u)+T(v)$ y $T(ku) = kT(u)$):
$$T(B(t)) = \sum_{i=0}^3 b_i(t) T(P_i)$$
Esto demuestra que la curva transformada sigue siendo una curva de Bézier, cuyos nuevos puntos de control son los originales transformados: $P'_i = T(P_i)$.

### c) Existencia de la transformación
Es posible definir $T$ si los puntos $P_0$ y $P_3$ forman una base de $\mathbb{R}^2$ (son linealmente independientes). En ese caso, existe una única matriz $A$ tal que:
$$A \cdot P_0 = Q_3 \quad \text{y} \quad A \cdot P_3 = Q_0$$

```python
import numpy as np
import matplotlib.pyplot as plt

def generar_puntos():
    return np.random.uniform(-10, 10, (4, 2))

def bezier_cubica(t, P):
    return (1-t)**3 * P[0] + 3*(1-t)**2*t * P[1] + 3*(1-t)*t**2 * P[2] + t**3 * P[3]

P = generar_puntos() # Puntos para B1
Q = generar_puntos() # Puntos para B2

# c) Hallar la matriz A de la transformación lineal
# Resolvemos A * [P0 | P3] = [Q3 | Q0]
try:
    orig = np.array([P[0], P[3]]).T  # Matriz con P0 y P3 como columnas
    dest = np.array([Q[3], Q[0]]).T  # Matriz con Q3 y Q0 como columnas

    A = dest @ np.linalg.inv(orig)   # A = dest * inv(orig)

    # Transformamos todos los puntos de control de B1
    P_trans = (A @ P.T).T

    # d) Gráficos
    t_vals = np.linspace(0, 1, 100)
    curva_b1 = np.array([bezier_cubica(t, P) for t in t_vals])
    curva_b2 = np.array([bezier_cubica(t, Q) for t in t_vals])
    curva_b1_t = np.array([bezier_cubica(t, P_trans) for t in t_vals])

    fig, axs = plt.subplots(1, 3, figsize=(18, 5))

    axs[0].plot(curva_b1[:,0], curva_b1[:,1], 'b'); axs[0].set_title('B1 Original')
    axs[1].plot(curva_b2[:,0], curva_b2[:,1], 'g'); axs[1].set_title('B2 Original')

    # Unión de T(B1) y B2
    axs[2].plot(curva_b1_t[:,0], curva_b1_t[:,1], 'b--', label='T(B1)')
    axs[2].plot(curva_b2[:,0], curva_b2[:,1], 'g', label='B2')
    axs[2].scatter([Q[3,0], Q[0,0]], [Q[3,1], Q[0,1]], color='black', zorder=5)
    axs[2].set_title('Unión: T(B1) + B2')
    axs[2].legend()
    plt.show()

except np.linalg.LinAlgError:
    print("Error: P0 y P3 son linealmente dependientes.")
```

Cada punto $B(t)$ de nuestra ruta es una **combinación lineal** de los cuatro puntos de control.

* **Optimización de Longitud**: Para cumplir con el objetivo de la **menor longitud posible**, buscamos que la combinación lineal de los puntos de control acerque la curva lo máximo posible a una línea recta, manteniendo el polígono de control en el límite de seguridad de los obstáculos para no ser descalificados.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle, Polygon

# 1. Definición de la función de Bézier Cúbica
def bezier_cubica(t, p0, p1, p2, p3):
    return (1-t)**3 * p0 + 3*(1-t)**2 * t * p1 + 3*(1-t) * t**2 * p2 + t**3 * p3

# 2. Puntos de control definidos
P0 = np.array([0, 0])
P1 = np.array([9.0, 0.25])
P2 = np.array([9.0, 0.5])
P3 = np.array([10, 5])

# 3. Generación de la curva (100 puntos)
t_vals = np.linspace(0, 1, 100)
curva = np.array([bezier_cubica(t, P0, P1, P2, P3) for t in t_vals])

# 4. Configuración del gráfico
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_aspect('equal')
ax.set_xlim(-1, 11)
ax.set_ylim(-1, 7)
ax.grid(True, alpha=0.3, linestyle='--')

# Dibujar Obstáculos Reales
pozo = Polygon([(1.5, 5.5), (2.2, 5.8), (2.8, 5.5), (2.5, 4.8), (1.8, 4.8)], facecolor='blue', alpha=0.5)
roca = Circle((4, 3), 1.2, facecolor='gray', alpha=0.6)
casa = Rectangle((6, 1), 2, 3, facecolor='brown', alpha=0.6)
arbustos = Polygon([(8.5, 6.2), (9.5, 6.0), (9.2, 5.3), (8.2, 5.5)], facecolor='green', alpha=0.4)

for patch in [pozo, roca, casa, arbustos]:
    ax.add_patch(patch)

# Dibujar Puntos de Inicio (A) y Destino (D)
ax.plot(0, 0, 'go', markersize=10)
ax.plot(10, 5, 'ro', markersize=10)

# Dibujar la Ruta y el Polígono de Control
ax.plot(curva[:, 0], curva[:, 1], 'm-', linewidth=3)
ax.plot([P0[0], P1[0], P2[0], P3[0]], [P0[1], P1[1], P2[1], P3[1]], 'yo--', alpha=0.4)



plt.title(f"Ejercicio 15")

plt.show()
```
